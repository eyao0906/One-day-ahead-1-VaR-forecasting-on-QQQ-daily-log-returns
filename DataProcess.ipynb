```python
import pandas as pd
import numpy as np
import os
import re
from pathlib import Path
from fileinput import filename
from matplotlib import lines
```

```python
PROJECT_ROOT = Path().resolve()
DATA_DIR = PROJECT_ROOT / "outputs"
garchx = pd.read_csv(DATA_DIR / "garchx_only/var_garchx_backtests.csv")
riskmetrics = pd.read_csv(DATA_DIR / "riskmetrics/var_riskmetrics_backtests.csv")
var = pd.read_csv(DATA_DIR / "var_backtests.csv")
var_ext = pd.read_csv(DATA_DIR / "var_backtests_ext.csv")



# Combine and remove exact duplicate rows
combined_df = pd.concat([garchx, riskmetrics, var, var_ext], 
                        ignore_index=True).drop_duplicates()

# Save
output_path = DATA_DIR / "Integrated_Backtest_Results.csv"
combined_df.to_csv(output_path, index=False)

print(f"Saved combined dataset to: {output_path}")
print(f"Final shape: {combined_df.shape}")
```
